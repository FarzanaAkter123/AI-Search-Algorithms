```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

# DFS Algorithm
def dfs(graph, start, goal):
    visited = set()
    traversal_order = []

    def dfs_recursive(node):
        if node in visited:
            return False

        visited.add(node)
        traversal_order.append(node)

        if node == goal:
            return True

        for neighbor in graph[node]:
            if dfs_recursive(neighbor):
                return True

        return False

    dfs_recursive(start)
    return traversal_order


# Run DFS
start_node = 'A'
goal_node = 'G'

result = dfs(graph, start_node, goal_node)

print("DFS Traversal:", result)
print("Goal Found:", goal_node in result)


# Visualization
G = nx.DiGraph(graph)

pos = {
    'A': (0, 2),
    'B': (-1, 1),
    'C': (1, 1),
    'D': (-1.5, 0),
    'E': (-0.5, 0),
    'F': (0.5, 0),
    'G': (1.5, 0)
}

plt.figure(figsize=(10, 6))

node_colors = []

for node in G.nodes():
    if node == start_node:
        node_colors.append('green')
    elif node == goal_node:
        node_colors.append('red')
    elif node in result:
        node_colors.append('skyblue')
    else:
        node_colors.append('lightgray')

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color=node_colors,
    node_size=2000,
    font_size=14,
    font_weight='bold',
    arrows=True
)

plt.title("Depth First Search (DFS) Visualization")
plt.show()
```
